````python

### 🟦 Célula 2 — Imports e lista de pratos
import os
import requests
import json

# Lista de pratos disponíveis
PRATOS = ["lasanha", "feijoada", "moqueca", "spaghetti alla carbonara", "strogonoff", "bobó de camarão"]
````

🟦 Célula 3 --- Intenção "ingredientes"

````python
# Frases que indicam intenção "ingredientes"
INTENT_INGREDIENTES = [
    "ingredientes", "quais são os ingredientes", "o que tem no prato",
    "buscar ingredientes", "receita do prato", "preciso dos ingredientes de",
    "o que vai no", "como faz o prato", "do que é feito"
]

# Função para detectar intenção
def detect_intent(text: str) -> str:
    txt = text.lower()
    for kw in INTENT_INGREDIENTES:
        if kw in txt:
            return "ingredientes"
    return "outro"
````

🟦 Célula 4 — Escolha interativa de pratos

````python
# Função para escolha interativa de prato(s)
def escolha_prato_interativa():
    print("\n🍽️ Pratos disponíveis:")
    for i, p in enumerate(PRATOS, start=1):
        print(f"{i}. {p}")
    escolha = input("\nDigite o número do prato (ou vários separados por vírgula): ").strip()
    indices = [s.strip() for s in escolha.split(",") if s.strip().isdigit()]
    selecionados = []
    for idx in indices:
        i = int(idx) - 1
        if 0 <= i < len(PRATOS):
            selecionados.append(PRATOS[i])
    if not selecionados:
        print("❌ Seleção inválida. Tente novamente.")
        return escolha_prato_interativa()
    return selecionados
````

🟦 Célula 5 — Consulta à API Deepseek

````python
# Função para consultar a API Deepseek
def consulta_deepseek(nome_prato: str, api_key: str):
    url = "https://openrouter.ai/api/v1/chat/completions"
    headers = {
        "Authorization": f"Bearer {api_key}",
        "Content-Type": "application/json",
    }
    prompt = f"Quais são os ingredientes necessários para preparar o prato chamado '{nome_prato}'?"
    payload = {
        "model": "deepseek/deepseek-r1:free",
        "messages": [
            {"role": "user", "content": prompt}
        ]
    }
    try:
        resp = requests.post(url, headers=headers, json=payload, timeout=20)
        j = resp.json()
        content = j.get("choices", [{}])[0].get("message", {}).get("content")
        return content or json.dumps(j, ensure_ascii=False, indent=2)
    except Exception:
        return f"⚠️ Erro na resposta da API: {resp.text}"
````

````python
🟦 Célula 6 — Execução principal
````

````python
# Execução principal
def main():
    user_input = input("Ingredientes: ")
    intent = detect_intent(user_input)
    if intent == "ingredientes":
        pratos_escolhidos = escolha_prato_interativa()
        api_key = os.getenv("DEEPSEEK_API_KEY") or input("🔑 Insira sua API key da Deepseek: ").strip()
        for prato in pratos_escolhidos:
            print(f"\n🔍 Buscando ingredientes para: {prato}\n")
            resultado = consulta_deepseek(prato, api_key)
            print(resultado)
    else:
        print("🤖 Intenção não reconhecida. Tente com frases relacionadas a ingredientes.")

# Rodar
main()
````
